#### TASK: Draw an empty RED rectangle around the dogs face and display the image in the notebook.

```python
dog_detected = img_rgb.copy()
cv2.rectangle(
    dog_detected,
    pt1=(195, 370),
    pt2=(610, 720),
    color=(255, 0, 0),
    thickness=5
)
plt.imshow(dog_detected)
```

#### TASK: Draw a BLUE TRIANGLE in the middle of the image. The size and angle is up to you, but it should be a triangle (three sides) in any orientation.

```python
import numpy as np

points = np.array([
    [400, 385],
    [200, 700],
    [600, 700]
], np.int32).reshape(-1, 1, 2)
```

```python
points.shape
```

```python
dog_triangle = img_rgb.copy()
cv2.polylines(
    dog_triangle,
    [points],
    isClosed=True,
    color=(0, 0, 255),
    thickness=5
)
plt.imshow(dog_triangle)
```

### BONUS TASK. Can you figure our how to fill in this triangle? It requires a different function that we didn't show in the lecture! See if you can use google search to find it.

[CLICK ME FOR A DIRECT LINK TO THE HINT](https://docs.opencv.org/3.0-beta/modules/imgproc/doc/drawing_functions.html#fillpoly)

```python
dog_triangle = img_rgb.copy()
cv2.fillPoly(
    dog_triangle,
    [points],
    color=(0, 0, 255)
)
plt.imshow(dog_triangle)
```

#### TASK: (NOTE: YOU WILL NEED TO RUN THIS AS A SCRIPT). Create a script that opens the picture and allows you to draw empty red circles where you click the RIGHT MOUSE BUTTON DOWN.

```python
img_mouse = img.copy()

def draw_cicle(event, x, y, flags, param):
    if event == cv2.EVENT_RBUTTONDOWN:
        cv2.circle(img_mouse, (x, y), 100, (0, 0, 255), 4)

cv2.namedWindow("Draw circle")
cv2.setMouseCallback("Draw circle", draw_cicle)

while True:
    cv2.imshow("Draw circle", img_mouse)
    if cv2.waitKey(1) & 0xFF == 27:
        break

cv2.destroyAllWindows()
```
